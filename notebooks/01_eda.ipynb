# 01 - Load and Inspect the Data

This notebook is only for the **first step** of the project:

1. Load the CSV  
2. Look at the first few rows  
3. Check shape and column names  
4. Check data types  
5. Check missing values  
6. Check duplicates  
7. Inspect the target column  
8. View summary statistics


```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)
pd.set_option("display.float_format", lambda x: f"{x:.2f}")
```

```python
file_path = "../data/heart_failure_clinical_records_dataset.csv"
df = pd.read_csv(file_path)
df.head()
```

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
print("Missing values per column:")
print(df.isnull().sum())
```

```python
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)
```

```python
print("Target value counts:")
print(df["DEATH_EVENT"].value_counts())

print("\nTarget proportions:")
print(df["DEATH_EVENT"].value_counts(normalize=True))
```

```python
df.describe()
```

```python
binary_columns = ["anaemia", "diabetes", "high_blood_pressure", "sex", "smoking", "DEATH_EVENT"]

for col in binary_columns:
    print(f"{col}: {sorted(df[col].unique().tolist())}")
```
